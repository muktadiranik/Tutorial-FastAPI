```python
import chromadb

client = chromadb.Client()

# Delete collection
client.delete_collection("news_articles")
```

```python
collection = client.create_collection("news_articles")
```

```python
collection.add(
    ids=["id1", "id2", "id3", "id4", "id5", "id6", "id7", "id8", "id9", "id10"], 
    documents=[
        "Apple is leading in a smart phone game with iPhone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C, and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $",
        "Apple's iPhone sales up by 35% in Q2",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C, and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $",
        "Apple's iPhone sales up by 35% in Q2",
        "Tesla booked a minor profit of 1 billion $ in Q2",
    ]
)
```

```python
results = collection.query(
    query_texts=["Elon Musk", "Tesla", "SpaceX"],
    n_results=1,
)
print(results)
```

```python
data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

for i, embedding in enumerate(data["embeddings"]):
    print(f"Embedding {i}: {embedding}")
    print(f"Metadata {i}: {data['metadatas'][i]}")
    print(f"Document {i}: {data['documents'][i]}")
    print()
```
